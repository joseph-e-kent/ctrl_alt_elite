```python
# Examine the data
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
# from matplotlib import ax

# 1. Define a helper function to format the numbers
def k_formatter(x, pos):
    # Divide by 1,000 and format with 0 or 1 decimal places as needed
    # Example: 1500000 becomes '1500k', 150000 becomes '150k'
    return f"{x * 1e-3:.0f}k"

SHOWALL = False

if SHOWALL:
    sorted_data = raw_data.sort_values(by="price", ascending=True) # Sort to ensure the large numbers are on top of the plot
    sorted_data["price"] = sorted_data["price"] / 10
    plot = sns.scatterplot(data=sorted_data,
                    x='lat', 
                    y='long', 
                    hue='price',
                    # size='price',
                    palette="viridis",
                    alpha=0.3,
                    edgecolor=None,
                    # s=50
                   )
    
    # 2. Apply the formatter directly to the legend/colorbar
    # Because Seaborn handles continuous hues with a Colorbar, we use update_normal
    if plot.legend_:
        # If it generated a standard legend box
        labels = plot.legend_.get_texts()
        for label in labels:
            try:
                # Convert text to float, format it, and update
                val = float(label.get_text().replace(",", ""))
                label.set_text(k_formatter(val, None))
            except ValueError:
                pass  # Skips the title of the legend (e.g., 'price')
    else:
        # If Seaborn generated a continuous colorbar instead of a legend box
        # This grabs the colorbar and formats its tick marks
        cbar = plot.collections[0].colorbar
        if cbar:
            cbar.plot.yaxis.set_major_formatter(ticker.FuncFormatter(k_formatter))
    
    # 3. Set title of chart
    plt.title("Price vs. Latitude and Longitude", fontsize=14, fontweight="bold", pad=15)
    
    # Optional: Clean up the axis labels while you're at it
    plt.xlabel("Latitude", fontsize=11)
    plt.ylabel("Longitude", fontsize=11)
    
    plt.show()
    # raw_data[['lat', 'long']].to_csv('locations.csv', index=False)


### NEW CHART ###
if SHOWALL:
    plot = sns.lineplot(data=raw_data,
                 x='sqft_living',
                 y='price',
                        # alpha = 0.5,
                )
    
    sns.regplot(
        data=raw_data,
        x="sqft_living",
        y="price",
        scatter=False,  # Don't add extra scatter points
        color="red",  # Make the trend line stand out
        label="Trend Line",
    )
    
    plt.title("Square feet vs price", fontsize=14, fontweight="bold", pad=15)
    plot.set_yscale('log')
    plt.grid(True, which='both')
    plt.show()



### NEW CHART ZIP CODE and PRICE ###
if SHOWALL:
    sns.boxplot(data = raw_data,
                x='zipcode',
                y='price',
                showfliers=False
               )
    plt.show()


if SHOWALL:
    sns.boxplot(data = raw_data,
                x='waterfront',
                y='price',
               )
    plt.show()


sns.barplot(data = raw_data,
             x='bathrooms',
            y='price'
           )
plt.show()
```

```python
# Condition the data for the model

target_col = 'price'
Y = raw_data[target_col]

# Input: raw_data: dataframe with data
# Output: features to use for training
# Output: List of cols used
def condition_data(raw_data):
    # Data fixing possibilities
    # Fix the date to be an integer?
    
    drop_cols = [
        'id', 
        'date', 
        # 'bedrooms', 
        # 'bathrooms', 
        # 'sqft_living', 
        # 'sqft_lot',
        # 'floors', 
        # 'waterfront', 
        # 'view', 
        # 'condition', 
        # 'grade', 
        # 'sqft_above',
        # 'sqft_basement', 
        # 'yr_built', 
        # 'yr_renovated', 
        # # 'zipcode', 
        # # 'lat', 
        # # 'long',
        # 'sqft_living15', 
        # 'sqft_lot15', 
    ]
    
    # Set the filtered data
    # raw_data['zipcode'] = raw_data['zipcode'].astype('category')
    
    data_filtered = raw_data.drop(columns=drop_cols)
    try:
        data_filtered = data_filtered.drop(columns=[target_col])
    except KeyError:
        pass

    # Eliminate errors in the dataset:
    data_filtered = data_filtered[data_filtered['bathrooms'] != 0]
    data_filtered = data_filtered[data_filtered['bedrooms'] != 0]
    
    # Pull out our target feature

    display(data_filtered.shape)
    
    # Features that we will use from the data.
    features = list(data_filtered.columns)
    # display(features)
    X = data_filtered[features]
    # X = data_filtered
    return X, features
```

```python
import sklearn as skl

X, features = condition_data(raw_data)


X_train, X_test, Y_train, Y_test = skl.model_selection.train_test_split(X, Y, test_size=0.2, random_state=8264)

# Y_train = np.log1p(Y_train)

X.columns
```

```python
## TRAIN SINGLE MODEL

import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

# Initialize the XGBoost Regressor
clf = xgb.XGBRegressor(
    n_estimators=212,       # Number of gradient boosted trees
    learning_rate=0.1,      # Step size shrinkage used to prevent overfitting
    max_depth=11,            # Maximum depth of a tree
    n_jobs=6,
    random_state=8264,
    objective='reg:absoluteerror',
    # objective='reg:quantileerror', quantile_alpha=0.5
    # enable_categorical=True, tree_method="hist"
)

# Train the model
clf.fit(X_train, Y_train)
print("Model training complete!")

# Make predictions
# Y_pred = np.expm1(clf.predict(X_test))
Y_pred = (clf.predict(X_test))

# Calculate metrics
# Calculate traditional metrics
mae = mean_absolute_error(Y_test, Y_pred)
rmse = root_mean_squared_error(Y_test, Y_pred)

# 1. Calculate the absolute percentage error for each prediction
# (We add a tiny value like 1e-8 to avoid division by zero if an actual value is exactly 0)
absolute_percentage_errors = np.abs((Y_test - Y_pred) / (Y_test + 1e-8))

# 2. Calculate what percentage of guesses fall within 5%, 10%, and 20% thresholds
within_5pct = np.mean(absolute_percentage_errors <= 0.05) * 100
within_10pct = np.mean(absolute_percentage_errors <= 0.10) * 100
within_20pct = np.mean(absolute_percentage_errors <= 0.20) * 100

# Print the traditional metrics
print(f"Mean Absolute Error (MAE): ${mae:.2f}")
print(f"Root Mean Squared Error (RMSE): ${rmse:.2f}")

print("-" * 40)
# Print your new intuitive accuracy metrics
print("--- Prediction Accuracy Thresholds ---")
print(f"Guesses within  5% of actual: {within_5pct:.2f}%")
print(f"Guesses within 10% of actual: {within_10pct:.2f}%")
print(f"Guesses within 20% of actual: {within_20pct:.2f}%")
print("-" * 40)

# Optional: Set a new record baseline based on the 10% threshold
# (e.g., if you previously got 75% of guesses within 10%, your record is 75.0)
pct_5_record = 33

if within_5pct > pct_5_record:
    print("THIS MODEL IS A NEW RECORD FOR 5% ACCURACY!")
else:
    print("This model did not beat the percentage record.")

import matplotlib.pyplot as plt

# Plot feature importance
xgb.plot_importance(clf)
plt.show()
```
