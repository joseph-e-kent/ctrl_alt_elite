```python
# Read suplimental data:
import pandas as pd

sup_data = pd.read_csv('SPD_Crime_Data__2008-Present_20260522.csv')

try:
    sup_data.empty
except NameError:
    sup_data = pd.read_csv('SPD_Crime_Data__2008-Present_20260522.csv')

# Clean up suplimental data:
# Keep rows where NO column contains 'REDACTED'
sup_data = sup_data[~(sup_data == 'REDACTED').any(axis=1)]

# Make report_year col for filtering
sup_data['report_year'] = sup_data['Report DateTime'].str[0:4].astype(int)


# Only keep data for the area we care about:
sup_data = sup_data[
        (sup_data["Latitude"].astype(float) <= raw_data.lat.max().astype(float))
        & (sup_data["Latitude"].astype(float) >= raw_data.lat.min().astype(float))
    ]
sup_data = sup_data[
        (sup_data["Longitude"].astype(float) <= raw_data.long.max().astype(float))
        & (sup_data["Longitude"].astype(float) >= raw_data.long.min().astype(float))
    ]
# Only keep data where the lat and lonare not the same


display(sup_data.head())
display(sup_data.tail())
display(sup_data.describe())
display(sup_data.columns)


sup_filtered = sup_data[
    (sup_data["report_year"] >= 2014)
    & (sup_data["report_year"] <= 2015)
]
display(len(sup_filtered))
sup_filtered = sup_filtered.sample(5)
display(sup_filtered.head())
```

```python
# Condition the data for the model
import numpy as np

target_col = 'price'
# Y = raw_data[target_col]

# Input: raw_data: dataframe with data
# Output: features to use for training
# Output: List of cols used
def condition_data(raw_data):
    Y = []
    # Data fixing possibilities
    # Fix the date to be an integer?
    
    drop_cols = [
        'id', 
        'date', 
        # 'bedrooms', 
        # 'bathrooms', 
        # 'sqft_living', 
        # 'sqft_lot',
        # 'floors', 
        # 'waterfront', 
        # 'view', 
        # 'condition', 
        # 'grade', 
        # 'sqft_above',
        # 'sqft_basement', 
        # 'yr_built', 
        # 'yr_renovated', 
        # # 'zipcode', 
        # # 'lat', 
        # # 'long',
        # 'sqft_living15', 
        # 'sqft_lot15', 
    ]
    
    # Set the filtered data
    raw_data['zipcode'] = raw_data['zipcode'].astype('category')
    raw_data['waterfront'] = raw_data['waterfront'].astype('category')
    raw_data['sale_year'] = raw_data['date'].str[0:4].astype(int)
    max_year = raw_data.sale_year.max()
    min_year = raw_data.sale_year.min()


    # Apply the conditional logic
    raw_data["years_since_renovated"] = np.where(
        raw_data["yr_renovated"] == 0,
        max_year - raw_data["yr_built"],  # If True (never renovated)
        max_year - raw_data["yr_renovated"],  # If False (was renovated)
    )
    
    data_filtered = raw_data.drop(columns=drop_cols)

    # Eliminate errors in the dataset:
    data_filtered = data_filtered[
        (data_filtered["bedrooms"] > 0)
        & (data_filtered["bathrooms"] > 0)
        & ((data_filtered["sqft_living"] / data_filtered["bedrooms"]) >= 70)
    ]

    ## Merge the suplimentary data into our dataset
    sup_filtered = sup_data[
        (sup_data["report_year"] >= min_year)
        & (sup_data["report_year"] <= max_year)
    ]

    ## Extract return variables:
    try:
        Y = data_filtered[target_col]
        data_filtered = data_filtered.drop(columns=[target_col])
        print(f"NOTE: dropping '{target_col}'")
    except KeyError:
        print(f"WARNING: Did not find/drop '{target_col}'")

    features = list(data_filtered.columns)
    X = data_filtered[features]

    return X, Y, features, data_filtered

X, Y, features, data_filtered = condition_data(raw_data)
```

```python
## TRAIN SINGLE MODEL

import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

# Initialize the XGBoost Regressor
clf = xgb.XGBRegressor(
    n_estimators=200,       # Number of gradient boosted trees
    learning_rate=0.1,      # Step size shrinkage used to prevent overfitting
    max_depth=10,            # Maximum depth of a tree
    n_jobs=6,
    random_state=8264,

    objective='reg:quantileerror', 
    quantile_alpha=0.5,
    # objective='reg:absoluteerror',
    # objective='reg:quantileerror', quantile_alpha=0.5
    enable_categorical=True,
    tree_method="hist"
)
Y_train_log = np.log(Y_train)
# Train the model
clf.fit(X_train, Y_train_log)
print("Model training complete!")

# Make predictions
# Y_pred = np.expm1(clf.predict(X_test))
Y_pred = np.exp(clf.predict(X_test))

# Calculate metrics
# Calculate traditional metrics
mae = mean_absolute_error(Y_test, Y_pred)
rmse = root_mean_squared_error(Y_test, Y_pred)

# 1. Calculate the absolute percentage error for each prediction
# (We add a tiny value like 1e-8 to avoid division by zero if an actual value is exactly 0)
absolute_percentage_errors = np.abs((Y_test - Y_pred) / (Y_test + 1e-8))

# 2. Calculate what percentage of guesses fall within 5%, 10%, and 20% thresholds
within_5pct = np.mean(absolute_percentage_errors <= 0.05) * 100
within_10pct = np.mean(absolute_percentage_errors <= 0.10) * 100
within_20pct = np.mean(absolute_percentage_errors <= 0.20) * 100

# Print the traditional metrics
print(f"Mean Absolute Error (MAE): ${mae:.2f}")
print(f"Root Mean Squared Error (RMSE): ${rmse:.2f}")

print("-" * 40)
# Print your new intuitive accuracy metrics
print("--- Prediction Accuracy Thresholds ---")
print(f"Guesses within  5% of actual: {within_5pct:.2f}%")
print(f"Guesses within 10% of actual: {within_10pct:.2f}%")
print(f"Guesses within 20% of actual: {within_20pct:.2f}%")
print("-" * 40)

# Plot feature importance
xgb.plot_importance(clf)
plt.show()
```
